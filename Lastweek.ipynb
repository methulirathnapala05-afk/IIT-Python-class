```python
data = pd.Series([75, 82, None, 90, None, 68])
print(data)
```

```python
data.isna().sum()
```

```python
mean_value = data.mean()
filled = data.fillna(mean_value)
print(filled)
```

```python
dropped = data.dropna()
print(dropped)
```

```python
len(data)      # Original length
len(dropped)   # After dropping
```

```python
import pandas as pd

visits = pd.Series(
    [120, 150, 90, 200, 180, 220, 160],
    index=["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
)

print(visits)

```

```python
visits.sum()
```

```python
visits.idxmax()
```

```python
visits.mean()
```

```python
visits[visits > 150]
```

```python
visits.sort_values(ascending=False)
```

```python
visits.plot(title="Daily Website Visits")
```
