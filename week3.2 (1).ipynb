```python
name=input("enter name")
marks=int(input("enter marks"))
if (marks >= 75):
    print("name:",name,"marks:",marks,"Grade A")
elif (marks >=65):
    print("name:",name,"marks:",marks,"Grade B")
elif (marks >=50):
    print("name:",name,"marks:",marks,"Grade C")
else:
    print("name:",name,"marks:",marks,"Grade F")
```
