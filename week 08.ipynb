### How to debug

```python
try:
    print(x)
except:
    print("An error occored")
```

```python
try:
    num=int(input("enter number:")) ##runtime error

except ValueError:
       print("An error occored")
else:
    print("no error found")
finally:
        ("This always run" )
```
